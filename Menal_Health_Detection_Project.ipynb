```python
#we  want to have only top 10 countries and rest will be 'Others'

top_countries=df[df['Country']!='Other']['Country'].value_counts().index[:10].to_list()
top_countries
```

# Data Encoding

if we give male 0 and female 1 in age then the model might think female is more important than male hence we use one hot encoding.
But for columns like Stress level we know Very High> High> Medium> Low then we can use label encoding

```python
df.columns
```

```python
#dividing all the columns according to the type of tranformation we want to apply on them

#skewed_columns we need to apply log transformation to make them normally distributed.(becacuse skewed data)
#other_numeric_columns we don't need to apply log tranformation but maybe something else. 
#ordinal_columns we need to apply label encoding to convert them into numbers.
#normal_columns we need to apply one hot encoding to convert them into numbers.

skewed_columns=['Study_Hours']
other_numeric_columns=['Age','Avg_Daily_Usage_Hours','Daily_Unlocks','Physical_Activity_Hours','Sleep_Hours_Per_Night']

ordinal_columns=['Stress_Level']
normal_columns=['Gender','Academic_Level','Most_Used_Platform','Grouped_Country','Purpose_Of_Use']
feature_columns=skewed_columns+other_numeric_columns+ordinal_columns+normal_columns
```

```python
x=df[feature_columns]
y=df['Mental_Health_Score']

from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

# Building Pipelines for Data Tranformation

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer, StandardScaler, OrdinalEncoder, OneHotEncoder
#labelEncoder is used for target variable which is categorical and we need to convert it into numbers.
#OrdinalEncoder is used for input features which are categorical and we need to convert it into numbers.

#skewed features
skew_pipeline= Pipeline(steps=[
    ('log_transform',FunctionTransformer(np.log1p)),
    ('scale',StandardScaler())
])

#numeri features
plain_numeric_pipeline=Pipeline(steps=[
    ('scale',StandardScaler())
])

#ordinal features
ordinal_pipeline=Pipeline(steps=[
    ('encode',OrdinalEncoder(categories=[['Low','Medium','High','Very High']])),
])

#nominal features
nominal_pipeline=Pipeline(steps=[
    ('encode',OneHotEncoder(handle_unknown='ignore'))
])
```

```python
#column transformer will manage the pipeline

from sklearn.compose import ColumnTransformer

preprocessor=ColumnTransformer(transformers=[
#   ("name", konsi_pipeline, konsi_columns)
    ('skewed_pipeline', skew_pipeline, skewed_columns),
    ('plain_numeric_pipeline', plain_numeric_pipeline, other_numeric_columns),
    ('ordinal_pipeline', ordinal_pipeline, ordinal_columns),
    ('nominal_pipeline', nominal_pipeline, normal_columns)
])
preprocessor
```

# Building Pipeline for Model

```python
#Linear Regression Model
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
#we are using mean_squared_error, mean_absolute_error, r2_score beacuse we are doing regression problem.
#we can not use accuracy_score, precision_score, recall_score, f1_score because they are used for classification problem.

lr_pipeline=Pipeline(steps=[
    ('preprocessor',preprocessor),
    ('regressor',LinearRegression())
]) 

lr_pipeline.fit(x_train,y_train)
lr_preds=lr_pipeline.predict(x_test)
lr_preds_train=lr_pipeline.predict(x_train)

lr_r2_testing=r2_score(y_test,lr_preds)
lr_r2_training=r2_score(y_train,lr_preds_train)

print(f"R2 score for testing data: {lr_r2_testing}")
print(f"R2 score for training data: {lr_r2_training}")

lr_mae=mean_absolute_error(y_test,lr_preds)

print(f"Mean Absolute Error for testing data: {lr_mae}")
```

```python
#Random Forest Model
from sklearn.ensemble import RandomForestRegressor

rf_pipeline=Pipeline(steps=[
    ('preprocessor',preprocessor),
    ('random forest',RandomForestRegressor(random_state=42))
])

rf_pipeline.fit(x_train,y_train)

rf_preds=rf_pipeline.predict(x_test)
rf_preds_train=rf_pipeline.predict(x_train)

rf_r2_testing=r2_score(y_test,rf_preds)
rf_r2_training=r2_score(y_train,rf_preds_train)

print(f"R2 score for testing data: {rf_r2_testing}")
print(f"R2 score for training data: {rf_r2_training}")

rf_mae=mean_absolute_error(y_test,rf_preds)

print(f"Mean Absolute Error for testing data: {rf_mae}")
```

```python
#Hyperparameter tuning for Random Forest Model

from sklearn.model_selection import RandomizedSearchCV

param_grid={
    'random forest__n_estimators':[100,200,400],  
    #"random forest__" means that the n_estimators will directly go to the "random forest" named process
    #else we will be thrown an error because n_estimator can not go in preprocessing
    'random forest__max_depth':[5,10,20,30],
    'random forest__min_samples_split':[2,5,10],
    'random forest__min_samples_leaf':[1,2,4],
}

random_search=RandomizedSearchCV(
    estimator=rf_pipeline,  #we want the rf_pipeline to be tuned as it has the RnadomForestRegressor model in it.
    param_distributions=param_grid,
    n_iter=15,
    cv=5,
    scoring='r2', #not accuracy because we are doing regression problem and not categorical problem.
    random_state=42,
    n_jobs=-1 #when -1 multiprocessing when 1 singleprocessing
)

random_search.fit(x_train,y_train)

```

```python
random_search.best_params_ 
#this will give the best parameters for the Random Forest Model after hyperparameter tuning.
```

```python
#tuned Random Forest Pipeline Model
rf_tuned_pipeline=random_search.best_estimator_
rf_tuned_preds=rf_tuned_pipeline.predict(x_test)
rf_tuned_preds_train=rf_tuned_pipeline.predict(x_train)

print(f'R2_Score for Random Forest Model after hyperparameter tuning: {r2_score(y_test,rf_tuned_preds)}')
print(f'R2_Score for Random Forest Model after hyperparameter tuning on training data: {r2_score(y_train,rf_tuned_preds_train)}')
print(f'Mean Absolute Error for Random Forest Model after hyperparameter tuning: {mean_absolute_error(y_test,rf_tuned_preds)}')
```

# Save The Model

```python
from sklearn.metrics import root_mean_squared_error

#for linearRegression Model
lr_rmse= root_mean_squared_error(y_test,lr_preds)

#for RandomForest Model
rf_rmse= root_mean_squared_error(y_test,rf_preds)

#for Tuned RandomForest Model
rf_tuned_training_preds=rf_tuned_pipeline.predict(x_train)
rf_tuned_rmse= root_mean_squared_error(y_test,rf_tuned_preds)
rf_tuned_r2_training=r2_score(y_train,rf_tuned_training_preds)
rf_tuned_r2_testing=r2_score(y_test,rf_tuned_preds)
rf_tuned_mae=mean_absolute_error(y_test,rf_tuned_preds)

#creating DataFrame to store the results of all the models
model_results=pd.DataFrame({
    'Model':['Linear Regression','RandomForest','Tuned RandomForest' ],
    'R2 Score':[lr_r2_testing,rf_r2_testing,rf_tuned_r2_testing],
    "Training R2 Score":[lr_r2_training,rf_r2_training,rf_tuned_r2_training],
    "Mean Absolute Error":[lr_mae,rf_mae,rf_tuned_mae],
    "Root Mean Absolute Error":[root_mean_squared_error(y_test,lr_preds),rf_rmse,rf_tuned_rmse]
})

model_results
```
